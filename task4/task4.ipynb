```python
df["Day"] = df["Order_Date"].dt.day
df["Weekday"] = df["Order_Date"].dt.day_name()
```

```python
numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount_Pct",
    "Gross_Sales",
    "Discount_Amount",
    "Net_Sales",
    "Profit",
    "Profit_Margin_Pct",
    "Delivery_Days",
    "Customer_Rating"
]
```

```python
Q1 = df["Net_Sales"].quantile(0.25)
Q3 = df["Net_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Net_Sales"] < lower) |
    (df["Net_Sales"] > upper)
]

print("Number of outliers:", len(outliers))
```

```python
yearly_sales = df.groupby("Year")["Net_Sales"].sum()

plt.figure(figsize=(8,5))
yearly_sales.plot(kind="bar")
plt.title("Total Net Sales by Year")
plt.xlabel("Year")
plt.ylabel("Net Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales = df.groupby("Month")["Net_Sales"].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Net Sales Trend")
plt.xlabel("Month")
plt.ylabel("Net Sales")
plt.show()
```

```python
region_sales = df.groupby("Region")["Net_Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Net Sales by Region")
plt.xlabel("Region")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales = df.groupby("Category")["Net_Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(9,5))
category_sales.plot(kind="bar")
plt.title("Net Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(9,5))
category_profit.plot(kind="bar")
plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
segment_sales = df.groupby("Customer_Segment")["Net_Sales"].sum()

plt.figure(figsize=(8,5))
segment_sales.plot(kind="bar")
plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Net Sales")
plt.xticks(rotation=30)
plt.show()
```

```python
channel_sales = df.groupby("Sales_Channel")["Net_Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
channel_sales.plot(kind="bar")
plt.title("Sales by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Net Sales")
plt.xticks(rotation=30)
plt.show()
```

```python
payment_counts = df["Payment_Method"].value_counts()

plt.figure(figsize=(8,5))
payment_counts.plot(kind="bar")
plt.title("Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Discount_Pct",
    y="Profit"
)

plt.title("Discount Percentage vs Profit")
plt.xlabel("Discount Percentage")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Delivery_Days"],
    bins=9,
    kde=True
)

plt.title("Distribution of Delivery Days")
plt.xlabel("Delivery Days")
plt.ylabel("Number of Orders")
plt.show()
```

```python
status_counts = df["Order_Status"].value_counts()

plt.figure(figsize=(8,5))
status_counts.plot(kind="bar")
plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)
plt.show()
```

```python
return_by_category = df.groupby("Category")["Return_Flag"].mean() * 100

print(return_by_category.sort_values(ascending=False))
```

```python
plt.figure(figsize=(9,5))

return_by_category.sort_values(ascending=False).plot(kind="bar")

plt.title("Return Rate by Product Category")
plt.xlabel("Category")
plt.ylabel("Return Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
top_products = (
    df.groupby("Product")["Net_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
plt.figure(figsize=(10,6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Net Sales")
plt.xlabel("Net Sales")
plt.ylabel("Product")

plt.show()
```
